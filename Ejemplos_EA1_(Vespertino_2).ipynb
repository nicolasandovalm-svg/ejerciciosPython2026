# Ejemplo 1

La tarifa de agua $T$ depende del consumo de agua $x$. Para calcularla se considera un cargo fijo de 750 pesos, más 1200 pesos por cada metro cúbico consumido.

La función asociada a esto es:

$$T(x) = 750 + 1200x$$

Determinar:

1) $T(0), T(2), T(8)$
2) Graficar la función

```python
#Definir la función, y luego de return escribir la función:
def T(x):
    return 750 + 1200*x

#Obtener imágenes de un valor "a" => print (función(a))
print("T(0)=",T(0))
print("T(2)=",T(2))
print("T(8)=",T(8))

```

```python
#Librerías para graficar
import matplotlib.pyplot as plt
import numpy as np

#Tener la función definida con "def"
def T(x):
    return 750 + 1200*x

#Se crea un arreglo numérico (Otorga los valores a la  Var indep) np.arange(ini,fin,paso)
x =np.arange(0,30,0.1)

#Graficar:
plt.plot(x,T(x) )   #plt.plot(x,funcion a graficar(x))
plt.title('Tarifa del agua en funcion del consumo')   #título
plt.xlabel('conumo de agua (m3)')      #Eje x
plt.ylabel('Tarifa (pesos)')       #Eje y
plt.grid()       #Cuadrícula
plt.show()
```

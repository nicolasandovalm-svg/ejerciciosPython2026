# Problema 1

Una red transmite datos a $100$ megabits por segundo. Crea una función utilizando *Python* que permita calcular la cantidad de datos transmitidos en una cierta cantidad de tiempo (en segundos). Luego utiliza la función y calcula cuántos datos se transmiten en:

* $45$ segundos
* $1,5$ minutos
* $1$ hora



```python
def datos(x):
    return 100*x

#1 min = 60 s; 1 hora = 3600 segundos
print("Con 45 s, se transmiten",datos(45),"megabits")
print("Con 1,5 min, se transmiten",datos(1.5*60),"megabits")
print("Con 1 hr, se transmiten",datos(1*3600),"megabits")
```

# Problema 6

La temperatura de un servidor (en °C) puede ser modelada mediante la función

$$T(t) = -0,5 t^{2} + 3 t + 20$$

donde $t$ corresponde al tiempo transcurrido (en horas) en un día laboral. Considere que la jornada laboral comienza a las 08:00 horas y termina a las 17:00 horas.

1.  Defina variable dependiente e independiente, indicando unidad de medida.
2.  Determine el dominio contextualizado del problema.
3.  Grafique la función utilizando *Python*.
4.  Mediante un análisis gráfico, estime cuándo el servidor alcanza la máxima temperatura y calcule cuánto es.
5.  Determine la temperatura del servidor a las 13:00 horas y al finalizar la jornada laboral.


*Pregunta 1*
* Variable dependiente: Temperatura del servidor (°C)
* Variable independiente: Tiempo (horas)

*Pregunta 2*
* Dominio T: $[0,9]$

```python
import matplotlib.pyplot as plt
import numpy as np

#Tener la función definida con "def"
def T(t):
    return -0.5 * t**2 + 3*t + 20

#Se crea un arreglo numérico (Otorga los valores a la  Var indep) np.arange(ini,fin,paso)
x =np.arange(0,9.1,0.1)

#Graficar:
plt.plot(x,T(x) )   #plt.plot(x,funcion a graficar(x))
plt.title('Temperatura de un servidor en función del tiempo')   #título
plt.xlabel('Tiempo (horas)')      #Eje x
plt.ylabel('Temperatura (°C)')       #Eje y
plt.xticks(np.arange(0,10,1))
plt.yticks(np.arange(7.5,25,1))
plt.grid()       #Cuadrícula
plt.show()
```

*Pregunta 4*

El máximo de temperatura se produce luego de 3 horas (11:00), y tiene un valor de 24,5°C

```python
#Pregunta 5
def T(t):
    return -0.5 * t**2 + 3*t + 20
print(T(5))
print(T(9))
```

A las 13 hrs la temperatura es de 22,5°C, y a las 17 hrs la temperatura es de 6,5°C

**Pregunta extra**

¿A que hora el servidor alcanza 22,5°C?

```python
import numpy as np
from scipy.optimize import fsolve

#Definir la función de la ecuación igualada a 0
def f(x):
    return T(x) - 22.5
#Ingresar np.linspace(ini,fin, n soluciones a buscar)
xo = np.linspace(0,9,2)
solucion = fsolve(f,xo)
print(solucion)
```

El servidor alcanza una temperatura de 22,5°C transcurrida 1 hora (09:00) y 5 horas (13:00)

# Problema 7

El número de usuarios de una red social se puede modelar mediante la función:

$$U(t) = \frac{1.000}{1+9e^{-0.5t}}$$

donde $t$ corresponde al tiempo transcurrido en meses.

1.  Defina variable dependiente e independiente, indicando unidad de medida.
2.  Determine la cantidad de usuarios transcurridos $12$ meses.
3.  Grafique la función, utilizando la librería *Matplotlib*, para los primeros dos años de funcionamiento.
4.  ¿Cuánto tiempo debe pasar para que la red social llegue a $800$ usuarios?

*Pregunta 1*
* Variable dependiente: Número de usuarios (unidades)
* Variable independiente: tiempo (meses)

```python
import numpy as np 
def U(t):
    return 1000 / (1 + 9*np.exp(-0.5*t))
print("2) Pasados 12 meses hay",round(U(12)),"usuarios")
```

```python
import numpy as np
import matplotlib.pyplot as plt

def U(t):
    return 1000 / (1 + 9*np.exp(-0.5*t))
x = np.arange(0, 24.1, 0.1)
plt.plot(x, U(x), label = 'N° usuarios')
plt.title('Número de usuarios en función del tiempo')
plt.xlabel('Tiempo (meses)')
plt.ylabel('N° usuarios (unidades)')
plt.xticks(np.arange(0,25,2))
plt.yticks(np.arange(0,1001,100))
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
from scipy.optimize import fsolve
def U(t):
    return 1000 / (1 + 9*np.exp(-0.5*t))

#Definir la nueva función igualando a cero
def f(x):
    return U(x)- 800
#Valor/es inicial/es de la aproximación
x = np.linspace(0,10,1)
solucion = fsolve(f, x)
print(f'El valor de x es {round(solucion[0],2)}')
```

Hay 800 usuarios luego de 7,17 meses

$$A(n) = 0,01n^{2} + 0,5n + 2$$

donde $n$ corresponde a la cantidad de elementos a ordenar. El algoritmo funciona desde las 23:00 horas hasta las 06:00 horas y debe ordenar $1.562$ elementos.

1. Defina variable dependiente e independiente, indicando unidad de medida.
2. Determine el dominio contextualizado de la función.
3. Determine el tiempo de ejecución para ordenar $1.200$ elementos.
4. Grafique, utilizando la biblioteca *Matplotlib*, la función $A(n)$.
5. ¿Cuántos elementos ordena luego de $6$ horas de funcionamiento?

* Variable dependiente: Tiempo (segundos)
* Variable independiente: Cantidad de elementos (unidades)

* Dominio: $[0, 1562]$

```python
def A(n):
    return 0.01 * n**2 + 0.5 * n + 2
print("El tiempo de ejecución de 1200 elementos es", A(1200), "segundos")
#1 hora = 3600 seg
```

```python
import numpy as np
import matplotlib.pyplot as plt
def A(n):
    return 0.01 * n**2 + 0.5 * n + 2
x = np.arange(0, 1563, 1)
plt.plot(x, A(x))
plt.title('Tiempo de ordenamiento en función del n° de elementos de un algoritmo')
plt.xlabel('N° elementos (unidades)')
plt.ylabel('Tiempo(segundos)')
plt.grid(True)
#plt.legend()
plt.show()
```

```python
import numpy as np
from scipy.optimize import fsolve
def A(n):
    return 0.01 * n**2 + 0.5 * n + 2
#Definir la nueva función igualando a cero
def f(x):
    return A(x)- 6*3600

#Valor/es inicial/es de la aproximación
x = np.linspace(0,10,1)
solucion = fsolve(f, x)
print(f'El valor de x es {round(solucion[0])}')
print("En 6 horas logra ordenar 1445 elementos aprox.")
```
